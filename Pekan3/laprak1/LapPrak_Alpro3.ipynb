## **2.1 Dasar Teori**
Operator merupakan simbol atau kata kunci khusus yang digunakan untuk mengolah nilai atau variabel (operand) guna menghasilkan nilai baru. Dalam bahasa pemrograman Python, operator dikelompokkan menjadi tujuh kategori utama berdasarkan fungsi dan penggunaannya:


##### **2.1.1 Operator Aritmatika**
   Operator aritmatika digunakan untuk melakukan operasi matematika pada Python. Operator ini meliputi penjumlahan (+), pengurangan (-), perkalian (*), pembagian (/), pembagian bulat (//), sisa bagi (%), dan perpangkatan (**).

##### **2.1.2 Operator Penugasan**
   Operator penugasan digunakan untuk memberikan atau mengubah nilai suatu variabel, yaitu
   = → Memberikan nilai, += → Menambahkan nilai,-= → Mengurangi nilai,*= → Mengalikan nilai,/= → Membagi nilai,//= → Pembagian bulat,%= → Sisa bagi,**= → Pangkat.

##### **2.1.3 Operator Perbandingan**
Operator perbandingan digunakan untuk membandingkan dua nilai dan menghasilkan nilai True atau False. Operator yang digunakan antara lain lebih besar (>), lebih kecil (<), lebih besar atau sama dengan (>=), lebih kecil atau sama dengan (<=), sama dengan (==), dan tidak sama dengan (!=).

##### **2.1.4 Operator Logika**
Operator logika digunakan untuk menggabungkan atau membalik kondisi yang menghasilkan True atau False.

and → True jika semua kondisi benar

or → True jika salah satu kondisi benar

not → Membalik nilai True menjadi False atau sebaliknya.

##### **2.1.5 Operator Bitwise**
Operator Bitwise Operator bitwise bekerja pada tingkat representasi biner (bit) dari suatu bilangan bulat. Operator ini meliputi AND (&), OR (|), XOR (^), NOT (~ atau dapat direpresentasikan dengan operasi negatif), geser kiri (<<), dan geser kanan (>>). Operator geser kiri akan mengalikan nilai dengan 2 pangkat jumlah pergeseran, sedangkan geser kanan akan membaginya.

##### **2.1.6 Operator Keanggotaan**
Operator keanggotaan digunakan untuk memeriksa apakah suatu nilai terdapat di dalam sebuah objek seperti string, list, tuple, atau set.

in → Memeriksa apakah nilai terdapat di dalam objek

not in → Memeriksa apakah nilai tidak terdapat di dalam objek.

##### **2.1.7 Operator Identitas**
Operator identitas digunakan untuk memeriksa apakah dua variabel merujuk pada objek yang sama.

is → Memeriksa apakah objek sama

is not → Memeriksa apakah objek berbeda

### **2.2.1 Program "aritmatika_2611532011.py"**


```python
angka1_2011 = int(input("Input angka-1: "))
angka2_2011 = int(input("Input angka-2: "))

#Penjumlahan
hasil_2011 = angka1_2011 + angka2_2011
print("\nOperator Penjumlahan")
print("Hasil =", hasil_2011)

#Pengurangan
hasil_2011 = angka1_2011 - angka2_2011
print("\nOperator Pengurangan")
print("Hasil =", hasil_2011)

#Perkalian
hasil_2011 = angka1_2011 * angka2_2011
print("\nOperator Perkalian")
print("Hasil =", hasil_2011)

#Pembagian, pembagian bulat, dan sisa bagi
if angka2_2011 != 0:
    hasil_2011 = angka1_2011 / angka2_2011
    print("\nOperator Pembagian")
    print("Hasil =", hasil_2011)

    hasil_2011 = angka1_2011 // angka2_2011
    print("\nOperator Pembagian Bulat")
    print("Hasil =", hasil_2011)

    hasil_2011 = angka1_2011 % angka2_2011
    print("\nOperator Sisa Bagi")
    print("Hasil =", hasil_2011)
else:
    print("Angka kedua tidak bernilai 0.")

 #Pangkat
hasil_2011 = angka1_2011 ** angka2_2011
print("\nOperator Pangkat")
print("Hasil =", hasil_2011)
```

**Penjelasan :**

* **Input angka pertama**

   `angka1_2011` digunakan untuk menyimpan angka pertama yang dimasukkan pengguna.`int(input(...))` digunakan agar input diubah menjadi bilangan bulat. begitu juga dengan `angka2_2011`


* **Operator Penjumlahan (`+`)**

  `angka1_2011 + angka2_2011` digunakan untuk menjumlahkan kedua angka.
  Hasilnya disimpan dalam variabel `hasil_2011` dan ditampilkan ke layar.

* **Operator Pengurangan (`-`)**

   `angka1_2011 - angka2_2011` digunakan untuk mengurangi angka pertama dengan angka kedua. Hasil pengurangan kemudian ditampilkan.

* **Operator Perkalian (`*`)**

   `angka1_2011 * angka2_2011` digunakan untuk mengalikan kedua angka.
   Hasil perkalian ditampilkan ke layar.

* **Operator Pembagian (`/`)**

   `angka1_2011 / angka2_2011` digunakan untuk membagi angka pertama dengan angka kedua.
   Sebelum melakukan pembagian, program memeriksa apakah angka kedua bukan 0.

* **Operator Pembagian Bulat (`//`)**

   Digunakan untuk membagi dua angka dan mengambil hasil bilangan bulatnya.


* **Operator Sisa Bagi (`%`)**

   Digunakan untuk mendapatkan sisa dari pembagian.

* **Percabangan `if`**

   `if angka2_2011 != 0:` memastikan angka kedua tidak bernilai 0.
   Hal ini dilakukan agar program tidak mengalami kesalahan saat melakukan pembagian.
   Jika angka kedua bernilai 0, program menampilkan pesan bahwa angka kedua tidak boleh 0.

* **Operator Pangkat (`**`)**

   `angka1_2011 ** angka2_2011` digunakan untuk menghitung pangkat.
   



### **2.2.2 Program "assignment_2611532011.py"**

```python
angka1_2011 = int(input("Input angka-1: "))
angka2_2011 = int(input("Input angka-2: "))

print("\nNilai awal angka1 =", angka1_2011)
print("Nilai angka2 =", angka2_2011)

#Assignment biasa
hasil_2011 = angka1_2011
print("\nAssignment Biasa (=)")
print("Hasil =", hasil_2011)

#Assignment penambahan
hasil_2011 = angka1_2011
hasil_2011 += angka2_2011
print("\nAssignment Penambahan (+=)")
print("Hasil =", hasil_2011)

#Assignment pengurangan
hasil_2011 = angka1_2011
hasil_2011 -= angka2_2011
print("\nAssignment Pengurangan (-=)")
print("Hasil =", hasil_2011)

#Assignment perkalian
hasil_2011 = angka1_2011
hasil_2011 *= angka2_2011
print("\nAssignment Perkalian (*=)")
print("Hasil =", hasil_2011)

#Assignment pembagian, pembagian bulat, dan sisa bagi
if angka2_2011 != 0:
    hasil_2011 = angka1_2011
    hasil_2011 /= angka2_2011
    print("\nAssignment pembagian (/=)")
    print("Hasil =", hasil_2011)
    #Operator tambahan
    hasil_2011 = angka1_2011
    hasil_2011 //= angka2_2011
    print("\nAssignment pembagian bulat (//=)")
    print("Hasil =", hasil_2011)
    hasil_2011 = angka1_2011
    hasil_2011 %= angka2_2011
    print("\nAssignment sisa bagi (%=)")
    print("Hasil =", hasil_2011)
else:
    print("\nPembagian tidak dapat dilakukan")
    print("Angka kedua tidak boleh bernilai 0.")

#Operator tambahanb: Assignment perpangkatan
hasil_2011 = angka1_2011
hasil_2011 **= angka2_2011
print("\nAssignment perpangkatan (**=)")
print("Hasil =", hasil_2011)
```

**Penjelasan:**

* **Input angka**
  Program meminta pengguna memasukkan dua angka melalui `input()`. Fungsi `int()` digunakan untuk mengubah input menjadi bilangan bulat dan disimpan dalam `angka1_2011` dan `angka2_2011`.

* **Menampilkan nilai awal**
  Program menampilkan nilai awal dari `angka1_2011` dan `angka2_2011` sebagai informasi sebelum dilakukan operasi assignment.

* **Assignment biasa (`=`)**
  Operator `=` digunakan untuk memberikan nilai `angka1_2011` ke variabel `hasil_2011`. Jadi, nilai `hasil_2011` akan sama dengan angka pertama.

* **Assignment penambahan (`+=`)**
  Operator `+=` digunakan untuk menambahkan `angka2_2011` ke `hasil_2011`. Operasi ini sama dengan `hasil_2011 = hasil_2011 + angka2_2011`.

* **Assignment pengurangan (`-=`)**
  Operator `-=` digunakan untuk mengurangi nilai `hasil_2011` dengan `angka2_2011`. Hasilnya kemudian disimpan kembali ke `hasil_2011`.

* **Assignment perkalian (`*=`)**
  Operator `*=` digunakan untuk mengalikan `hasil_2011` dengan `angka2_2011`. Hasil perkalian disimpan kembali ke variabel tersebut.

* **Assignment pembagian (`/=`)**
  Operator `/=` digunakan untuk membagi `hasil_2011` dengan `angka2_2011`. Program terlebih dahulu memastikan angka kedua tidak bernilai `0`.

* **Assignment pembagian bulat (`//=`)**
  Operator `//=` digunakan untuk melakukan pembagian bulat. Hasil yang diperoleh hanya berupa bagian bilangan bulat dari pembagian.

* **Assignment sisa bagi (`%=`)**
  Operator `%=` digunakan untuk mendapatkan sisa hasil pembagian antara `hasil_2011` dan `angka2_2011`.

* **Percabangan `if`**
  Kondisi `if angka2_2011 != 0` digunakan untuk memastikan angka kedua bukan `0`. Jika angka kedua bernilai `0`, program menampilkan pesan bahwa pembagian tidak dapat dilakukan.

* **Assignment perpangkatan (`**=`)**
  Operator `**=` digunakan untuk menghitung perpangkatan. Nilai `hasil_2011` dipangkatkan dengan `angka2_2011`, kemudian hasilnya disimpan kembali.


### **2.2.3 Program "bitwise_2611532011.py"**

```python
print("\n===================")
print("3. OPERATOR BITWISE")
print("===================")

angka1_2011 = int(input("Masukkan angka bitwise-1: "))
angka2_2011 = int(input("Masukkan angka bitwise-2: "))

print("\nAngka dalam bentuk desimal dan biner")
print("angka1_2011 =", angka1_2011, "| biner", bin(angka1_2011))
print("angka2_2011 =", angka2_2011, "| biner", bin(angka2_2011))

# Bitwise AND
hasil_2011 = angka1_2011 & angka2_2011
print("\nBitwise AND (&)")
print(angka1_2011, "&", angka2_2011, "=", hasil_2011)
print("Biner hasil =", bin(hasil_2011))
print("Biner hasil (8 bit) =", format(hasil_2011, "08b"))

# Bitwise OR
hasil_2011 = angka1_2011 | angka2_2011
print("\nBitwise OR (|)")
print(angka1_2011, "|", angka2_2011, "=", hasil_2011)
print("Biner hasil =", bin(hasil_2011))
print("Biner hasil (8 bit) =", format(hasil_2011, "08b"))

# Bitwise XOR
hasil_2011 = angka1_2011 ^ angka2_2011
print("\nBitwise XOR (^)")
print(angka1_2011, "^", angka2_2011, "=", hasil_2011)
print("Biner hasil =", bin(hasil_2011))
print("Biner hasil (8 bit) =", format(hasil_2011, "08b"))

# Bitwise NOT
hasil_2011 = ~angka1_2011
print("\nBitwise NOT (~)")
print("~", angka1_2011, "=", hasil_2011)
print("Biner hasil =", bin(hasil_2011))
print("Biner hasil (8 bit) =", format(hasil_2011, "08b"))

# Bitwise geser kiri
jumlah_geser_2011 = int(input("\nMasukkan jumlah pergeseran bit: "))

hasil_2011 = angka1_2011 << jumlah_geser_2011
print("\nBitwise geser kiri (<<)")
print(angka1_2011, "<<", jumlah_geser_2011, "=", hasil_2011)
print("Biner hasil =", bin(hasil_2011))
print("Biner hasil (8 bit) =", format(hasil_2011, "08b"))

# Bitwise geser kanan
hasil_2011 = angka1_2011 >> jumlah_geser_2011
print("\nBitwise geser kanan (>>)")
print(angka1_2011, ">>", jumlah_geser_2011, "=", hasil_2011)
print("Biner hasil =", bin(hasil_2011))
print("Biner hasil (8 bit) =", format(hasil_2011, "08b"))
```

**Penjelasan:**

* **Input dan tampilan angka:** Program meminta pengguna memasukkan dua angka untuk digunakan dalam operasi bitwise. Nilai tersebut disimpan dalam `angka1_2011` dan `angka2_2011`, kemudian ditampilkan dalam bentuk desimal dan biner menggunakan fungsi `bin()`.

* **Bitwise AND (`&`):** Operator AND digunakan untuk membandingkan bit dari dua angka. Setiap bit akan menghasilkan `1` apabila kedua bit bernilai `1`, sedangkan jika salah satu atau keduanya bernilai `0`, hasilnya `0`. Hasil operasi kemudian ditampilkan dalam bentuk desimal dan biner.

* **Bitwise OR (`|`):** Operator OR membandingkan bit dari dua angka dan menghasilkan `1` jika salah satu atau kedua bit bernilai `1`. Hasilnya digunakan untuk melihat perubahan nilai setelah dilakukan operasi OR pada kedua angka.

* **Bitwise XOR (`^`):** Operator XOR menghasilkan `1` jika kedua bit memiliki nilai yang berbeda dan menghasilkan `0` jika kedua bit memiliki nilai yang sama. Operasi ini dilakukan pada setiap pasangan bit dari kedua angka yang dimasukkan.

* **Bitwise NOT (`~`):** Operator NOT digunakan untuk membalik nilai setiap bit pada suatu angka. Bit yang bernilai `1` akan menjadi `0`, sedangkan bit `0` menjadi `1`. Pada program ini, operasi NOT dilakukan pada `angka1_2011`.

* **Bitwise geser kiri (`<<`) dan geser kanan (`>>`):** Program meminta jumlah pergeseran bit dari pengguna. Operator `<<` digunakan untuk menggeser bit ke arah kiri, sedangkan `>>` digunakan untuk menggeser bit ke arah kanan. Operasi ini dapat mengubah nilai angka berdasarkan jumlah posisi bit yang digeser.

* **Format hasil biner:** Fungsi `format(hasil_2011, "08b")` digunakan untuk menampilkan hasil dalam bentuk biner dengan panjang 8 bit. Dengan demikian, hasil operasi bitwise menjadi lebih mudah diamati dan dibandingkan.


### **2.2.4 Program "lainnya_2611532011.py"**

```python
print("===================================")
print("1. OPERATOR KEANGGOTAAN")
print("===================================")

# Input beberapa data yang dipisahkan dengan koma
input_data_2011 = input("Masukkan beberapa angka, pisahkan dengan koma: ")

# Mengubah input menjadi list integer
data_2011 = [int(angka_2011.strip()) for angka_2011 in input_data_2011.split(",")]

nilai_dicari_2011 = int(input("Masukkan angka yang ingin dicari; "))

#Operator in
hasil_2011 = nilai_dicari_2011 in data_2011
print("\nOperator keanggotaan in")
print(nilai_dicari_2011, "in", data_2011, "=", hasil_2011)

#Operator Not In
hasil_2011 = nilai_dicari_2011 not in data_2011
print("\nOperator keanggotaan Not In")
print(nilai_dicari_2011, "not in", data_2011, "=", hasil_2011)

print("===================================")
print("2. OPERATOR IDENTITAS")
print("===================================")

# Objek 1 menggunakan list dari pengguna
objek1_2011 = data_2011

#objek 2 merujuk pada objek yang sama dengan objek 1
objek2_2011 = objek1_2011

#objek 3 memiliki isi sama, tetapi merupakan objek baru
objek3_2011 = data_2011.copy()

print("objek1_2011 =", objek1_2011)
print("objek2_2011 =", objek2_2011)
print("objek3_2011 =", objek3_2011)

# operator is
hasil_2011 = objek1_2011 is objek2_2011
print("\nOperator identitas is")
print("objek1_2011 is objek2_2011 = ", hasil_2011)

#operator is not
hasil_2011 = objek1_2011 is not objek3_2011
print("\nOperator identitas is not")
print("objek1_2011 is not objek3_2011 = ", hasil_2011)

# Membandingkan identitas dan nilai
print("\nPerbandingan identitas dan nilai:")
print("objek1_2011 is objek3_2011:", objek1_2011 is objek3_2011)
print("objek1_201 == objek3_2011:", objek1_2011 == objek3_2011)
```

**Penjelasan:**

* **Memasukkan data:** Program meminta pengguna memasukkan beberapa angka yang dipisahkan dengan koma. Data tersebut kemudian diubah menjadi bentuk list integer menggunakan `split()` untuk memisahkan angka dan `int()` untuk mengubahnya menjadi bilangan bulat.

* **Operator keanggotaan (`in` dan `not in`):** Program meminta satu angka yang ingin dicari di dalam list. Operator `in` digunakan untuk mengecek apakah angka tersebut terdapat dalam list, sedangkan `not in` digunakan untuk mengecek apakah angka tersebut tidak terdapat dalam list. Hasil pengecekan berupa `True` atau `False`.

* **Membuat objek:** `objek1_2011` berisi list yang telah dimasukkan pengguna. Kemudian `objek2_2011` dibuat dengan mengarah pada objek yang sama dengan `objek1_2011`, sedangkan `objek3_2011` dibuat menggunakan `.copy()`, sehingga memiliki isi yang sama tetapi merupakan objek yang berbeda.

* **Operator identitas (`is` dan `is not`):** Operator `is` digunakan untuk mengecek apakah dua variabel menunjuk pada objek yang sama. Karena `objek1_2011` dan `objek2_2011` menunjuk objek yang sama, hasilnya `True`. Sementara itu, `objek1_2011 is not objek3_2011` menghasilkan `True` karena keduanya merupakan objek yang berbeda.

* **Perbandingan identitas dan nilai:** Bagian terakhir menunjukkan perbedaan antara `is` dan `==`. Operator `is` membandingkan identitas objek, sedangkan `==` membandingkan isi atau nilai objek. Jadi, meskipun `objek1_2011` dan `objek3_2011` merupakan objek yang berbeda, keduanya tetap dapat menghasilkan `True` ketika dibandingkan menggunakan `==` karena memiliki isi yang sama.




### **Program "logika_2611532011.py"**

```python
a1_2011 = input("Input nilai boolean-1 (True/False): ").strip().lower() == "true"
a2_2011 = input("Input nilai boolean-2 (True/False): ").strip().lower() == "true"

print("\nA1 =", a1_2011)
print("A2 =", a2_2011)

 # Konjungsi: bernilai true jika keduanya True
hasil_2011 = a1_2011 and a2_2011
print("\nKonjungsi (AND)")
print("A1 and A2=", hasil_2011)

# Disjungsi: bernilai true jika salah satu True
hasil_2011 = a1_2011 or a2_2011
print("\nDisjungsi (OR)")
print("A1 or A2=", hasil_2011)

# Negasi A1: membalik nilai A1
hasil_2011 = not a1_2011
print("\nNegasi A1 (NOT)")
print("not A1=", hasil_2011)

# Negasi A2: membalik nilai A2
hasil_2011 = not a2_2011
print("\nNegasi A2 (NOT)")
print("not A2=", hasil_2011)

#XOR: bernilai true jika keduanya berbeda
hasil_2011 = a1_2011 != a2_2011
print("\nDisjungsi Ekslusif (XOR)")
print("A1 XOR A2=", hasil_2011)
```

**Penjelasan:**

* **Input nilai Boolean:** Program meminta pengguna memasukkan dua nilai, yaitu `True` atau `False`. Fungsi `.strip().lower()` digunakan untuk menghilangkan spasi dan mengubah tulisan menjadi huruf kecil, kemudian `== "true"` mengubah input menjadi nilai Boolean.

* **Operator AND (`and`):** Operator `and` digunakan untuk menggabungkan dua kondisi. Hasilnya akan `True` hanya jika A1 dan A2 sama-sama bernilai `True`. Jika salah satunya `False`, hasilnya menjadi `False`.

* **Operator OR (`or`):** Operator `or` menghasilkan `True` jika salah satu atau kedua nilai bernilai `True`. Hasilnya hanya `False` apabila A1 dan A2 keduanya `False`.

* **Operator NOT (`not`):** Operator `not` digunakan untuk membalik nilai Boolean. Jika nilai awal `True`, hasilnya menjadi `False`, dan jika nilai awal `False`, hasilnya menjadi `True`. Program melakukan negasi pada A1 dan A2.

* **Operator XOR:** Pada program ini, XOR dibuat menggunakan `!=`. Hasilnya `True` jika nilai A1 dan A2 berbeda, sedangkan jika keduanya sama, hasilnya `False`. Semua hasil operasi kemudian ditampilkan ke layar.

```python
angka1_2011 = int(input("Input angka-1: "))
angka2_2011 = int(input("Input angka-2: "))

#Lebih besar dari
hasil_2011 = angka1_2011 > angka2_2011
print("\nOperator Lebih Besar Dari")
print("Hasil =", hasil_2011)

#Lebih kecil dari
hasil_2011 = angka1_2011 < angka2_2011
print("\nOperator Lebih Kecil Dari")
print("Hasil =", hasil_2011)

# Lebih besar dari atau sama dengan
hasil_2011 = angka1_2011 >= angka2_2011
print("\nOperator Lebih Besar Dari atau Sama Dengan")
print("Hasil =", hasil_2011)

# Lebih kecil dari atau sama dengan
hasil_2011 = angka1_2011 <= angka2_2011
print("\nOperator Lebih Kecil Dari atau Sama Dengan")
print("Hasil =", hasil_2011)

# Sama dengan
hasil_2011 = angka1_2011 == angka2_2011
print("\nOperator Sama Dengan")
print("Hasil =", hasil_2011)

# Tidak sama dengan
hasil_2011 = angka1_2011 != angka2_2011
print("\nOperator Tidak Sama Dengan")
print("Hasil =", hasil_2011)

# Tambahan: perbandingan berantai dalam python
hasil_2011 = 0 < angka1_2011 < 100
print("\n Perbandingan Berantai")
print("0 < angka1 < 100 =", hasil_2011)

hasil_2011 = 0 < angka2_2011 < 100
print("0 < angka2 < 100 =", hasil_2011)
```

```python
# Program Studi Kasus Terpadu: Sistem Simulasi Transaksi dan Validasi Akses Toko
# Nama File: tugas3_2611532011.py
# Mata Kuliah: Praktikum Algoritma dan Pemrograman

print("=== SISTEM TRANSAKSI TOKO ===")

# 2. Data Pelanggan dan Transaksi (Menggunakan input)
nama_2011 = input("Masukkan Nama Pelanggan : ")
status_pelanggan_2011 = input("Masukkan Status Pelanggan (member/nonmember) : ")
total_belanja_2011 = int(input("Masukkan Total Belanja : "))
jumlah_barang_2011 = int(input("Masukkan Jumlah Barang : "))
kode_promo_2011 = input("Masukkan Kode Promo : ")

print("\n=== DATA TRANSAKSI ===")
print("Nama Pelanggan   :", nama_2011)
print("Status Pelanggan :", status_pelanggan_2011)
print("Total Belanja    : Rp", total_belanja_2011)
print("Jumlah Barang    :", jumlah_barang_2011)
print("Kode Promo       :", kode_promo_2011)

# 4. Operator Perbandingan
is_belanja_cukup_2011 = total_belanja_2011 >= 200000
is_barang_cukup_2011 = jumlah_barang_2011 >= 3
is_member_2011 = status_pelanggan_2011.strip().lower() == "member"

# 7. Operator Keanggotaan (Membership)
daftar_promo_2011 = ["HEMAT10", "HEMAT20", "GRATISONGKIR"]
is_promo_tersedia_2011 = kode_promo_2011.strip().upper() in daftar_promo_2011

# 5. Operator Logika
is_dapat_diskon_2011 = is_member_2011 and is_belanja_cukup_2011
is_dapat_promo_2011 = is_promo_tersedia_2011 and is_barang_cukup_2011
member_access_2011 = is_member_2011 or is_belanja_cukup_2011
promo_access_2011 = is_dapat_promo_2011
free_shipping_access_2011 = is_promo_tersedia_2011 or is_barang_cukup_2011

print("\n=== HASIL VALIDASI ===")
print("Belanja >= Rp200000   :", is_belanja_cukup_2011)
print("Jumlah Barang >= 3    :", is_barang_cukup_2011)
print("Status Member         :", is_member_2011)
print("Kode Promo Tersedia   :", is_promo_tersedia_2011)
print("Mendapatkan Diskon    :", is_dapat_diskon_2011)
print("Mendapatkan Promo     :", is_dapat_promo_2011)

# 3. Operator Aritmatika
if is_dapat_diskon_2011:
    diskon_2011 = int(total_belanja_2011 * 0.1)
else:
    diskon_2011 = 0

total_pembayaran_2011 = total_belanja_2011 - diskon_2011
rata_rata_2011 = total_belanja_2011 / jumlah_barang_2011 if jumlah_barang_2011 > 0 else 0

print("\n=== HASIL PERHITUNGAN ===")
print("Diskon                : Rp", diskon_2011)
print("Total Pembayaran      : Rp", total_pembayaran_2011)
print("Rata-rata Harga Barang: Rp", int(rata_rata_2011))

print("\n=== HAK AKSES PELANGGAN ===")
print("Member Access         :", member_access_2011)
print("Promo Access          :", promo_access_2011)
print("Free Shipping Access  :", free_shipping_access_2011)

# 6. Operator Penugasan (Assignment & Augmented Assignment)
poin_2011 = 100        # Assignment biasa
poin_2011 += 50        # Augmented assignment penambahan

# 8. Operator Identitas (Identity)
objek_a_2011 = ["Akses", "Toko"]
objek_b_2011 = objek_a_2011
objek_c_2011 = ["Akses", "Toko"]
cek_is_2011 = objek_a_2011 is objek_b_2011
cek_is_not_2011 = objek_a_2011 is not objek_c_2011

# 9. Operator Bitwise
# Nilai Bit: 0001 (Member), 0010 (Belanja >= 200k), 0100 (Jumlah barang >= 3), 1000 (Kode promo tersedia)
bit_member_2011 = 1 if is_member_2011 else 0
bit_belanja_2011 = 2 if is_belanja_cukup_2011 else 0
bit_barang_2011 = 4 if is_barang_cukup_2011 else 0
bit_promo_2011 = 8 if is_promo_tersedia_2011 else 0

kode_status_bit_2011 = bit_member_2011 | bit_belanja_2011 | bit_barang_2011 | bit_promo_2011

print("\n=== OPERASI BITWISE ===")
print("=== Kode Status Transaksi ===")
print("0001 | 0010 | 0100 | 1000")
print("Kode Biner  :", bin(kode_status_bit_2011)[2:].zfill(4))
print("Kode Desimal:", kode_status_bit_2011)

print("\n=== Pemeriksaan Status ===")
print("Cek Member")
print(f"{bin(kode_status_bit_2011)[2:].zfill(4)} & 0001")
hasil_and_member_2011 = kode_status_bit_2011 & 1
print("Hasil Biner :", bin(hasil_and_member_2011)[2:].zfill(4))
print("Hasil Desimal:", hasil_and_member_2011)

print("\nCek Promo")
print(f"{bin(kode_status_bit_2011)[2:].zfill(4)} & 1000")
hasil_and_promo_2011 = kode_status_bit_2011 & 8
print("Hasil Biner :", bin(hasil_and_promo_2011)[2:].zfill(4))
print("Hasil Desimal:", hasil_and_promo_2011)

print("\n=== Perbandingan Status ===")
kode_referensi_2011 = 0b1011  # 11 dalam desimal
print("Kode Transaksi :", bin(kode_status_bit_2011)[2:].zfill(4))
print("Kode Referensi : 1011")
hasil_xor_2011 = kode_status_bit_2011 ^ kode_referensi_2011
print(f"{bin(kode_status_bit_2011)[2:].zfill(4)} ^ 1011")
print("Hasil Biner :", bin(hasil_xor_2011)[2:].zfill(4))
print("Hasil Desimal:", hasil_xor_2011)

print("\n=== Shift ===")
hasil_shift_2011 = kode_status_bit_2011 << 1
print(f"{bin(kode_status_bit_2011)[2:].zfill(4)} << 1")
print("Hasil Biner :", bin(hasil_shift_2011)[2:].zfill(4))
print("Hasil Desimal:", hasil_shift_2011)

print("\n=== SELESAI ===")
```
