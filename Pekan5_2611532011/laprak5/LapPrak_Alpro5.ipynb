## **2.1 Dasar Teori**

**1. Pengertian Perulangan (Looping)**

Perulangan atau *looping* merupakan salah satu konsep dasar dalam pemrograman yang digunakan untuk menjalankan suatu perintah secara berulang. Perulangan sangat berguna ketika sebuah perintah harus dilakukan beberapa kali dengan proses yang sama. Dengan menggunakan perulangan, programmer tidak perlu menulis kode yang sama secara berulang-ulang sehingga program menjadi lebih singkat, rapi, dan mudah dipahami. Dalam bahasa pemrograman Python, terdapat beberapa jenis perulangan, salah satunya adalah for loop.

**2. Perulangan For pada Python**

Perulangan for adalah struktur kontrol yang digunakan untuk melakukan pengulangan terhadap setiap elemen yang terdapat dalam suatu kumpulan data atau urutan tertentu. Perulangan for dapat digunakan pada berbagai jenis data, seperti list, tuple, string, maupun angka yang dihasilkan menggunakan fungsi range(). Pada setiap proses pengulangan, variabel akan mengambil nilai dari elemen yang sedang diproses.


**3. Fungsi range()**

Fungsi range() merupakan fungsi bawaan Python yang digunakan untuk menghasilkan urutan angka. Fungsi ini sering digunakan bersama dengan perulangan for untuk menentukan berapa kali suatu perintah akan dijalankan. range() dapat digunakan dengan satu, dua, atau tiga parameter. Jika menggunakan satu parameter, angka tersebut menjadi batas akhir dan dimulai dari 0.

**4. Nested For Loop**

Nested for loop atau perulangan for bersarang adalah perulangan for yang berada di dalam perulangan for lainnya. Perulangan ini digunakan ketika suatu proses membutuhkan lebih dari satu tingkat pengulangan. Setiap kali perulangan luar berjalan satu kali, perulangan dalam akan menjalankan seluruh prosesnya. Nested for loop sering digunakan untuk membuat pola, mengolah data dalam bentuk baris dan kolom, serta melakukan pengulangan yang memiliki struktur lebih kompleks.

**5. Variabel dalam Perulangan For**

Variabel pada perulangan for berfungsi untuk menyimpan nilai dari setiap elemen yang sedang diproses. Nilai variabel tersebut dapat berubah pada setiap pengulangan sesuai dengan data yang diberikan.

**6. Penerapan Perulangan For**

Perulangan for dapat diterapkan dalam berbagai permasalahan pemrograman. Contohnya adalah menampilkan daftar data, menghitung jumlah atau total nilai, mencetak angka dalam urutan tertentu, membuat pola, dan mengolah data dalam jumlah banyak. Penggunaan perulangan membuat program lebih efisien karena satu blok kode dapat digunakan untuk memproses banyak data. Oleh karena itu, pemahaman terhadap perulangan for, fungsi range(), dan nested for loop sangat penting sebagai dasar dalam mempelajari pemrograman Python.

### **2.2.1 Program "nested_for1_2611532011.py"**


```python
batas_2011 = int(input("Masukkan nilai batas: "))
for line_2011 in range(1, batas_2011 + 1):
    for j_2011 in range(1, (-1 * line_2011 + batas_2011) + 1):
        print(".", end="")
    print(line_2011)
```

**Penjelasan :**

1. batas_2011 = int(input("Masukkan nilai batas: "))
Digunakan untuk meminta pengguna memasukkan nilai batas. input() menerima masukan dari pengguna, sedangkan int() mengubah masukan tersebut menjadi bilangan bulat.

2. for line_2011 in range(1, batas_2011 + 1):
Merupakan perulangan for bagian luar yang digunakan untuk menentukan baris. Perulangan dimulai dari angka 1 sampai nilai batas_2011.

3. for j_2011 in range(1, (-1 * line_2011 + batas_2011) + 1):
Merupakan perulangan for bagian dalam atau nested for. Perulangan ini menentukan jumlah tanda titik (.) yang akan dicetak pada setiap baris.

4. -1 * line_2011 + batas_2011
Digunakan untuk menghitung jumlah titik. Semakin besar nilai line_2011, jumlah titik akan semakin sedikit. Jadi, titik akan berkurang pada setiap baris.

5. print(".", end="")
Digunakan untuk mencetak tanda titik tanpa pindah ke baris baru. end="" membuat titik berikutnya tetap dicetak di baris yang sama.

6. print(line_2011)
Setelah semua titik pada satu baris selesai dicetak, program mencetak nomor baris. Karena print() secara default berpindah ke baris baru, hasil berikutnya akan muncul di baris selanjutnya.

### **2.2.4 Program "nested_for4_2611532011.py"**





```python
tinggi_2011 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_2011 % 2 !=0:
    print("Tinggi harus bilangan genap!")
else:
    a_2011 = tinggi_2011
    c_2011 = a_2011
    lebar_2011 = (2 * tinggi_2011) - 2

    for i_2011 in range(1, tinggi_2011 + 1):
        b_2011 = c_2011 + 1

        for j_2011 in range(1, lebar_2011 + 1):

            #Baris atas dan bawah
            if i_2011 == 1 or i_2011 == tinggi_2011:
                if j_2011 == 1 or j_2011 == lebar_2011:
                    print("#", end="")
                else:
                    print("=", end="")

            #Baris isi
            else:
                if j_2011 == 1 or  j_2011 == lebar_2011:
                    print("|", end="")
                else:
                    if j_2011 == c_2011:
                        print("<", end="")
                    elif j_2011 == b_2011:
                        print(">", end="")
                    elif j_2011 == (lebar_2011 - c_2011):
                        print("<", end="")
                    elif j_2011 == (lebar_2011 - c_2011 + 1):
                        print(">", end="")
                    elif j_2011 > b_2011 and j_2011< (lebar_2011 - c_2011):
                        print(".", end="")
                    else:
                        print(" ", end="")
        print()

        # Logika asli Java
        a_2011 -= 2

        if a_2011 <= 0:
            c_2011 = (-a_2011) +2
        else:
            c_2011 = a_2011
```

### **Program "perulangan_for3_2611532011.py"**

```python
ulang_2011 = int(input("Masukkan jumlah perulangan: "))

jumlah_2011 = 0
for i_2011 in range(1, ulang_2011 + 1):
    print(i_2011, end="")
    jumlah_2011 = jumlah_2011 + i_2011

    if i_2011 < ulang_2011:
        print(" + ", end="")
    else:
        print(" = ", jumlah_2011, end="")
print()
print("jumlah =", jumlah_2011)
```

Digunakan untuk melakukan perulangan dari angka 1 sampai jumlah perulangan yang dimasukkan.

4. print(i_2011, end="")
Menampilkan angka yang sedang diproses tanpa pindah ke baris baru.

5. jumlah_2011 = jumlah_2011 + i_2011
Menambahkan nilai i_2011 ke jumlah_2011 sehingga semua angka dapat dijumlahkan.

6. if i_2011 < ulang_2011:
Mengecek apakah angka yang sedang diproses belum menjadi angka terakhir.

7. print(" + ", end="")
Jika belum angka terakhir, program menampilkan tanda + sebagai pemisah.

8. else:
Jika sudah mencapai angka terakhir, program menjalankan perintah berikutnya.

9. print(" = ", jumlah_2011, end="")
Menampilkan tanda = dan hasil penjumlahan keseluruhan.

10. print()
Digunakan untuk berpindah ke baris berikutnya setelah proses selesai.

```python
# === PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ===

n_2011 = int(input("Masukkan ukuran skala jam pasir (N): "))

print("=== PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ===")

# Bingkai atas
print("#", end="")
for i_2011 in range(4 * n_2011 + 5):
    print("=", end="")
print("#")

# Fase 1: Jam pasir bagian atas
for baris_2011 in range(n_2011, 0, -1):

    print("|", end="")

    print(" ", end="")

    for spasi_2011 in range(2 * (n_2011 - baris_2011)):
        print(" ", end="")

    for angka_2011 in range(baris_2011, 0, -1):
        print(angka_2011, end=" ")

    print("<*>", end="")

    for angka_2011 in range(1, baris_2011 + 1):
        print(" ", end="")
        print(angka_2011, end="")

    for spasi_2011 in range(2 * (n_2011 - baris_2011)):
        print(" ", end="")

    print(" |")

# Fase 2: Titik pusat
print("|", end="")

for spasi_2011 in range(2 * n_2011 + 1):
    print(" ", end="")

print("<*>", end="")

for spasi_2011 in range(2 * n_2011 + 1):
    print(" ", end="")

print("|")

# Fase 3: Jam pasir bagian bawah
for baris_2011 in range(1, n_2011 + 1):

    print("|", end="")

    print(" ", end="")

    for spasi_2011 in range(2 * (n_2011 - baris_2011)):
        print(" ", end="")

    for angka_2011 in range(baris_2011, 0, -1):
        print(angka_2011, end=" ")

    print("<*>", end="")

    for angka_2011 in range(1, baris_2011 + 1):
        print(" ", end="")
        print(angka_2011, end="")

    for spasi_2011 in range(2 * (n_2011 - baris_2011)):
        print(" ", end="")

    print(" |")

# Bingkai bawah
print("#", end="")
for i_2011 in range(4 * n_2011 + 5):
    print("=", end="")
print("#")
```

for baris_2011 in range(n_2011, 0, -1) membuat baris dari ukuran terbesar menuju terkecil. Pada bagian ini terdapat:
Perulangan untuk membuat spasi di sebelah kiri.
Perulangan untuk mencetak angka secara menurun, misalnya 4 3 2 1.

6. Fase 3 – Jam pasir bawah

# **KESIMPULAN**

Berdasarkan praktikum yang telah dilakukan, dapat disimpulkan bahwa perulangan `for` pada bahasa pemrograman Python sangat berguna untuk menjalankan suatu perintah secara berulang sesuai dengan jumlah yang telah ditentukan. Melalui beberapa program yang dibuat, perulangan `for` dapat digunakan untuk menampilkan angka secara berurutan, menghitung jumlah bilangan, mencari dan menjumlahkan bilangan genap, serta membuat berbagai bentuk pola menggunakan karakter dan angka. Penggunaan `range()` juga membantu menentukan nilai awal, batas akhir, dan arah perulangan sehingga program dapat berjalan sesuai dengan kebutuhan. Selain itu, nested `for` dapat digunakan ketika program membutuhkan perulangan di dalam perulangan, seperti untuk membuat pola yang memiliki baris dan kolom.

Pada program jam pasir kristal palindromik, penggunaan nested `for` diterapkan untuk mengatur setiap bagian pola, mulai dari bingkai, spasi, angka yang menurun dan menaik, hingga simbol `<*>` sebagai bagian tengah pola. Nilai `N` yang dimasukkan oleh pengguna menentukan ukuran dari pola yang akan dibuat, sehingga program dapat menghasilkan bentuk yang berbeda tanpa harus menulis kode baru. Dari praktikum ini dapat dipahami bahwa perulangan tidak hanya digunakan untuk mengulang perintah, tetapi juga dapat membantu membuat program yang lebih terstruktur, efisien, dan mudah dikembangkan. Dengan memahami penggunaan `for`, `range()`, nested `for`, `if`, dan `end=""`, kita dapat membuat berbagai program dan pola dengan lebih mudah menggunakan Python.
