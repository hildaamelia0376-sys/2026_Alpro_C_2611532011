## **2.1 Dasar Teori**
##### **1. Pengertian umum Python**

Python adalah bahasa pemrograman tingkat tinggi yang digunakan untuk membuat berbagai macam program dan aplikasi. Python memiliki sintaks atau aturan penulisan yang sederhana dan mudah dipahami sehingga cocok digunakan oleh pemula untuk mempelajari dasar-dasar pemrograman.

Python mendukung berbagai konsep pemrograman, seperti variabel, tipe data, percabangan, perulangan, fungsi, dan struktur data. Dalam praktikum ini, Python digunakan untuk mempelajari cara menyimpan dan mengolah data menggunakan variabel serta berbagai tipe data seperti **integer, float, string, dan boolean**.


##### **2. Variabel dalam python**
   Variabel dalam Python adalah tempat untuk menyimpan data atau nilai yang dapat digunakan dan diolah kembali dalam sebuah program. Variabel dapat menyimpan berbagai jenis data, seperti angka, teks, maupun nilai benar atau salah. Dalam Python, variabel dibuat dengan cara memberikan nama variabel dan nilai menggunakan tanda sama dengan (=).

##### **3. Type data dasar**
   1. Integer (int)

        Integer adalah tipe data yang digunakan untuk menyimpan bilangan bulat, yaitu angka yang tidak memiliki nilai desimal.
2. String (str)

      String adalah tipe data yang digunakan untuk menyimpan teks atau kumpulan karakter. String ditulis menggunakan tanda petik tunggal (' ') atau tanda petik ganda (" ").
3. Float (float)

      Float adalah tipe data yang digunakan untuk menyimpan bilangan yang memiliki angka desimal.
4. Boolean (bool)

      Boolean adalah tipe data yang hanya memiliki dua nilai, yaitu True (benar) dan False (salah). Boolean biasanya digunakan untuk menentukan kondisi dalam sebuah program.
##### **4. Konstanta dalam bahasa python**
Konstanta adalah nilai yang seharusnya tidak berubah selama program dijalankan. Konstanta biasanya digunakan untuk menyimpan nilai tetap, seperti nilai π (pi), jumlah hari dalam seminggu, atau nilai pajak.

Berbeda dengan beberapa bahasa pemrograman lain, Python tidak memiliki kata kunci khusus untuk membuat konstanta. Biasanya, programmer menggunakan huruf kapital pada nama variabel sebagai tanda bahwa nilai tersebut dianggap sebagai konstanta.
##### **5. Input dan Output dalam python**
1. Input

    Input adalah proses memasukkan data ke dalam program. Dalam Python, input biasanya dilakukan menggunakan fungsi input(). Data yang dimasukkan oleh pengguna melalui input() secara default akan dibaca sebagai string.Karena input() menghasilkan string, jika ingin memasukkan angka dan melakukan perhitungan, kita perlu mengubah tipe datanya menggunakan int() atau float().
2. Output

    Output adalah proses menampilkan hasil dari program kepada pengguna. Dalam Python, output biasanya menggunakan fungsi print().

## **2.2 Kode Program**

### **2.2.1 Program "Numerik_2611532011.py"**


```python
a_2011=100
print("The type of variable having value", a_2011, "is", type(a_2011))

#create a variable with float value
b_2011=10.2345
print("The type of variable having value", b_2011, "is", type(b_2011))

#creat a variable with complex value
c_2011=100+3j
print("The type of variable having value", c_2011, "is", type(c_2011))
```

**Penjelasan :**

* `a_2011 = 100`  Membuat variabel `a_2011` dengan nilai 100 yang bertipe **integer (int)**.
* `type(a_2011)`  Digunakan untuk mengetahui tipe data dari variabel `a_2011`.
* `b_2011 = 10.2345` → Membuat variabel `b_2011` dengan nilai 10.2345 yang bertipe **float (float)**.
* `type(b_2011)`  Digunakan untuk mengetahui tipe data dari variabel `b_2011`.
* `c_2011 = 100 + 3j` → Membuat variabel `c_2011` dengan nilai 100 + 3j yang bertipe **complex (complex)**.
* `type(c_2011)`  Digunakan untuk mengetahui tipe data dari variabel `c_2011`.
* `print()`  Digunakan untuk menampilkan nilai dan tipe data setiap variabel.

### **2.2.3 Program "Konstanta_2611532011.py"**

```python
#Program ini menggunakan konstanta untuk menghitung luas lingkaran
#nama variable ditambah 4 digit nim terakhir contoh: jari_1234

from typing import Final
PI: Final = 3.14
print("pi: %f" % (PI))
jari_2011 = float(input("Masukan nilai jari jari: "))
luas_2011 = PI * jari_2011 * jari_2011
print("luas lingkaran dengan jari jari %.2f adalah %.2f" % (jari_2011, luas_2011))
```

**Penjelasan:**

* `PI: Final = 3.14`  Membuat konstanta `PI` dengan nilai 3.14.
* `print(...)`  Menampilkan nilai `PI`.
* `jari_2011 = float(input(...))`  Meminta pengguna memasukkan nilai jari-jari dan mengubahnya menjadi tipe data float.
* `luas_2011 = PI * jari_2011 * jari_2011`  Menghitung luas lingkaran menggunakan nilai π dan jari-jari.
* `print(...)`  Menampilkan hasil luas lingkaran dengan 2 angka di belakang koma.
* `from typing import Final`  Menandai `PI` sebagai konstanta yang seharusnya tidak diubah.


### **2.2.4 Program "Input_2611532011.py"**

```python
panjang_2011 = input("Masukan nilai panjang:")
lebar_2011 = input("Masukan nilai lebar:")
luas_2011 = int(panjang_2011) * int(lebar_2011)
print("Luas =", luas_2011)
```

**Penjelasan:**

* `panjang_2011 = input(...)`  Meminta pengguna memasukkan nilai panjang.
* `lebar_2011 = input(...)`  Meminta pengguna memasukkan nilai lebar.
* `int(panjang_2011)` dan `int(lebar_2011)`  Mengubah data input menjadi tipe integer agar dapat dihitung.
* `luas_2011 = int(panjang_2011) * int(lebar_2011)`  Menghitung luas dengan rumus panjang * lebar.
* `print("Luas =", luas_2011)`  Menampilkan hasil perhitungan luas.




### **Program "Boolean_2611532011.py"**

```python
is_lulus = True
is_cumlaude = True

#Menggunakan Boolean
nilai_2011 = 85
batas_lulus = 75

#Menentukan nilai Boolean dari kondisi
status_kelulusan = nilai_2011 >= batas_lulus #Hasilnya akan true

print("=== Check Kelulusan ===")
print("Nilai:", nilai_2011)
print("Apakah Lulus?:", status_kelulusan)
if is_lulus and is_cumlaude:
    print("Selamat, Anda Lulus dengan predikat Cumlaude!")
```

**Penjelasan:**


* `is_lulus = True`  Membuat variabel Boolean yang menyatakan status lulus bernilai benar.
* `is_cumlaude = True`  Membuat variabel Boolean yang menyatakan status cumlaude bernilai benar.
* `nilai_2011 = 85`  Menyimpan nilai mahasiswa yaitu 85.
* `batas_lulus = 75`  Menentukan nilai minimum untuk dinyatakan lulus, yaitu 75.
* `status_kelulusan = nilai_2011 >= batas_lulus`  Membandingkan nilai 85 dengan batas lulus 75. Hasilnya True karena 85 >= 75.
* `print(...)`  Menampilkan informasi nilai dan status kelulusan.
* `if is_lulus and is_cumlaude:`  Mengecek apakah kedua kondisi bernilai True menggunakan operator `and`.
* `print("Selamat...")`  Menampilkan ucapan bahwa pengguna lulus dengan predikat Cumlaude jika kedua kondisi terpenuhi.
