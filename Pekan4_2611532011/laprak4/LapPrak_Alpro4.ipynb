## **2.1 Dasar Teori**

   **2.1.1. Conditional**

Conditional atau percabangan adalah struktur dalam pemrograman yang digunakan untuk membuat keputusan berdasarkan kondisi tertentu. Dengan conditional, program dapat menjalankan perintah yang berbeda sesuai dengan hasil dari suatu kondisi. Dalam Python, conditional dapat diterapkan menggunakan beberapa struktur seperti `if`, `if-else`, `if-elif-else`, `multi if`, dan `match-case`.

**2.1.2. If Tunggal**

`if` tunggal merupakan bentuk percabangan paling sederhana dalam Python. Struktur ini digunakan untuk menjalankan suatu perintah apabila kondisi yang diberikan bernilai benar (True). Jika kondisi bernilai salah (False), perintah di dalam `if` tidak akan dijalankan dan program akan melanjutkan ke perintah berikutnya.

**2.1.3. If-Else**

`if-else` digunakan ketika terdapat dua kemungkinan kondisi. Jika kondisi pada `if` bernilai benar, maka perintah pada bagian `if` akan dijalankan. Sebaliknya, jika kondisi tersebut bernilai salah, program akan menjalankan perintah pada bagian `else`. Struktur ini sering digunakan untuk menentukan dua pilihan, seperti lulus atau tidak lulus.

**2.1.4. If-Elif-Else**

`if-elif-else` digunakan ketika program memiliki lebih dari dua kemungkinan kondisi. `if` digunakan untuk memeriksa kondisi pertama, sedangkan `elif` digunakan untuk memeriksa kondisi berikutnya jika kondisi sebelumnya tidak terpenuhi. Jika semua kondisi `if` dan `elif` tidak terpenuhi, maka perintah pada `else` akan dijalankan. Struktur ini dapat digunakan untuk menentukan kategori nilai, seperti A, B, C, atau D.

**2.1.5. Multi If**

Multi `if` adalah penggunaan beberapa pernyataan `if` secara terpisah dalam satu program. Setiap kondisi akan diperiksa secara independen sehingga lebih dari satu kondisi dapat terpenuhi dan dijalankan. Berbeda dengan `if-elif-else`, pada multi `if` program tidak berhenti setelah menemukan satu kondisi yang benar, tetapi tetap memeriksa kondisi `if` berikutnya.

**2.1.6. Match-Case**

`match-case` merupakan struktur percabangan pada Python yang digunakan untuk mencocokkan suatu nilai dengan beberapa kemungkinan pola atau pilihan. Struktur ini memiliki fungsi yang mirip dengan `switch-case` pada beberapa bahasa pemrograman lain. `match` digunakan untuk menentukan nilai yang akan diperiksa, sedangkan `case` digunakan untuk menentukan kemungkinan nilai yang sesuai. `case _` dapat digunakan sebagai kondisi umum apabila tidak ada `case` yang cocok.

**2.1.7. Struktur Percabangan dalam Python**

Struktur percabangan membantu program mengambil keputusan berdasarkan kondisi atau nilai tertentu. Pemilihan jenis percabangan dapat disesuaikan dengan kebutuhan program. `if` tunggal cocok untuk satu kondisi, `if-else` untuk dua kemungkinan, `if-elif-else` untuk beberapa kondisi yang saling berkaitan, multi `if` untuk kondisi yang diperiksa secara terpisah, sedangkan `match-case` cocok untuk mencocokkan satu nilai dengan beberapa pilihan.

### **2.2.1 Program "if_elif_else1_2611532011.py"**


```python
umur_2011 = int(input("Input umur anda: "))
sim_2011 = input("Apakah Anda Sudah Punya SIM C: ")[0]

if umur_2011 >= 17 and sim_2011 == 'y':
    print("Anda sudah dewasa dan boleh bawa motor")
elif umur_2011 >= 17 and sim_2011 != 'y':
     print("Anda sudah dewasa tetapi tidak boleh bawa motor")
elif umur_2011 < 17 and sim_2011 =='y':
     print("Anda belum cukup umur punya SIM")
else:
     print("Anda belum cukup umur dan tidak boleh bawa motor")
print("Program Selesai")
```

**Penjelasan :**

1. **`umur_2011 = int(input("Input umur anda: "))`**
   Digunakan untuk memasukkan umur. Nilai input diubah menjadi tipe data `integer` agar dapat digunakan dalam perbandingan angka.

2. **`sim_2011 = input("Apakah Anda Sudah Punya SIM C: ")[0]`**
   Digunakan untuk memasukkan jawaban mengenai kepemilikan SIM C. `[0]` mengambil karakter pertama dari input, misalnya `y` dari jawaban `yes`.

3. **`if umur_2011 >= 17 and sim_2011 == 'y':`**
   Memeriksa apakah umur pengguna minimal 17 tahun dan memiliki SIM C. Jika kedua kondisi benar, program menampilkan bahwa pengguna sudah dewasa dan boleh membawa motor.

4. **`elif umur_2011 >= 17 and sim_2011 != 'y':`**
   Memeriksa pengguna yang berumur minimal 17 tahun tetapi tidak memiliki SIM C. Program akan menampilkan bahwa pengguna sudah dewasa tetapi tidak boleh membawa motor.

5. **`elif umur_2011 < 17 and sim_2011 == 'y':`**
   Memeriksa pengguna yang berumur di bawah 17 tahun tetapi menjawab memiliki SIM C. Program menampilkan bahwa pengguna belum cukup umur untuk memiliki SIM.

6. **`else:`**
   Digunakan jika semua kondisi sebelumnya tidak terpenuhi, yaitu ketika pengguna belum berusia 17 tahun dan tidak memiliki SIM C.

7. **`print("Program Selesai")`**
   Menampilkan tulisan “Program Selesai” setelah seluruh proses percabangan selesai dijalankan.

### **2.2.4 Program "match_case1_2611532011.py"**





```python
bulan_2011 = int(input("Masukkan Angka Bulan (1-12): "))

match bulan_2011:
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka Tidak Valid")
```

**Penjelasan:**

1. **`bulan_2011 = int(input("Masukkan Angka Bulan (1-12): "))`**
   Digunakan untuk memasukkan angka bulan dari 1 sampai 12. Fungsi `int()` mengubah input menjadi tipe data integer agar dapat dicocokkan dengan nilai pada `case`.

2. **`match bulan_2011:`**
   Digunakan untuk memeriksa nilai yang terdapat pada variabel `bulan_2011`. Struktur `match` akan mencocokkan nilai tersebut dengan beberapa pilihan `case`.

3. **`case 1:` sampai `case 12:`**
   Setiap `case` mewakili angka bulan tertentu. Jika angka yang dimasukkan sesuai dengan salah satu `case`, program akan menjalankan perintah `print()` yang sesuai.

4. **`case _:`**
   Digunakan sebagai kondisi umum (default) apabila nilai yang dimasukkan tidak sesuai dengan `case` yang tersedia. Misalnya, jika memasukkan angka `13`, program akan menampilkan “Angka Tidak Valid”.

5. **`print()` pada setiap case**
   Digunakan untuk menampilkan nama bulan berdasarkan angka yang dimasukkan.

### **Program "multi_if2_2611532011.py"**

```python
# Input dari user
total_belanja_2011 = float(input("Masukkan Total Belanja (Rp): "))

# Input status member (mengecek apakah user mengetik 'y' atau 'ya')
input_member_2011 = input("Apakah Anda Member? (y/t): ").strip().lower()
is_member_2011 = input_member_2011 in ["y", "ya"]

# Input status kode promo (mengecek apakah user mengetik 'y' atau 'ya')
input_promo_2011 = input("Apakah Kode Promo Valid (y/t): ").strip().lower()
kode_promo_valid_2011 = input_promo_2011 in ["y", "ya"]

total_diskon_persen_2011 = 0

# Multi-IF terpisah: Setiap kondisi diperiksa secara independen
# Diskon bisa ditumpuk (akumulasi) jika memenuhi beberapa syarat sekaligus
if total_belanja_2011 > 1000000:
    total_diskon_persen_2011 += 10 # Diskon belanja besar

if is_member_2011:
    total_diskon_persen_2011 += 5 # Diskon belanja member

if kode_promo_valid_2011:
    total_diskon_persen_2011 += 15 # Diskon belanja voucher

# Menghitung nominal diskon dan total bayar
nominal_diskon_2011 = total_belanja_2011 * (total_diskon_persen_2011 / 100)
total_bayar_2011 = total_belanja_2011 - nominal_diskon_2011

# Output hasil
print("\n--- Rincian Pembayaran ---")
print(f"Total Diskon : {total_diskon_persen_2011}% (Rp {nominal_diskon_2011:,.0f})")
print(f"Total Bayar  : Rp {total_bayar_2011:,.0f}")

print(f"Total Diskon yang Anda Dapatkan: {total_diskon_persen_2011}%")
# Output: Total diskon yang anda dapatkan: 30% jika belanja > 1 juta, member, dan kode promo valid
```

####

**Penjelasan:**

1. **`total_belanja_2011 = float(input(...))`**
   Digunakan untuk memasukkan total belanja. Fungsi `float()` digunakan agar nilai belanja dapat berupa angka desimal.

2. **`input_member_2011 = input(...).strip().lower()`**
   Digunakan untuk meminta status member. `strip()` menghapus spasi di awal atau akhir input, sedangkan `lower()` mengubah input menjadi huruf kecil.

3. **`is_member_2011 = input_member_2011 in ["y", "ya"]`**
   Digunakan untuk memeriksa apakah pengguna memasukkan `y` atau `ya`. Jika sesuai, variabel `is_member_2011` bernilai `True`.

4. **`input_promo_2011 = input(...).strip().lower()`**
   Digunakan untuk meminta pengguna memasukkan status kode promo.

5. **`kode_promo_valid_2011 = input_promo_2011 in ["y", "ya"]`**
   Digunakan untuk memeriksa apakah kode promo dinyatakan valid berdasarkan jawaban pengguna.

6. **`total_diskon_persen_2011 = 0`**
   Digunakan untuk menentukan nilai awal total diskon sebesar **0%** sebelum program memeriksa berbagai kondisi.

7. **`if total_belanja_2011 > 1000000:`**
   Memeriksa apakah total belanja lebih dari Rp1.000.000. Jika terpenuhi, program menambahkan 10% diskon menggunakan `+= 10`.

8. **`if is_member_2011:`**
   Memeriksa apakah pengguna merupakan member. Jika benar, program menambahkan 5% diskon.

9. **`if kode_promo_valid_2011:`**
   Memeriksa apakah kode promo valid. Jika benar, program menambahkan 15% diskon.

10. **Penggunaan multi-if**
     Ketiga kondisi tersebut menggunakan **multi-if**, sehingga setiap kondisi diperiksa secara terpisah.

11. **`nominal_diskon_2011 = total_belanja_2011 * (total_diskon_persen_2011 / 100)`**
      Digunakan untuk menghitung jumlah uang yang mendapatkan potongan berdasarkan persentase diskon yang diperoleh.

12. **`total_bayar_2011 = total_belanja_2011 - nominal_diskon_2011`**
     Digunakan untuk menghitung jumlah yang harus dibayar setelah dikurangi nominal diskon.

13. **`print("\n--- Rincian Pembayaran ---")`**
     Menampilkan judul bagian hasil pembayaran agar output program lebih rapi.

14. **`print(f"Total Diskon : ...")`**
     Menampilkan persentase total diskon dan nominal uang yang didapatkan dari potongan harga.

15. **`print(f"Total Bayar : ...")`**
     Menampilkan jumlah akhir yang harus dibayar setelah mendapatkan diskon.



### **Program "nilai_2611532011.py"**

2. **`if nilai_2011 >= 81:`**
   Memeriksa apakah nilai yang dimasukkan 81 atau lebih. Jika kondisi terpenuhi, program menampilkan nilai huruf A.

3. **`elif nilai_2011 >= 70:`**
   Jika kondisi `if` sebelumnya tidak terpenuhi, program memeriksa apakah nilai 70 atau lebih. Jika benar, program menampilkan B.

4. **`elif nilai_2011 >= 60:`**
   Memeriksa apakah nilai 60 atau lebih setelah kondisi sebelumnya tidak terpenuhi. Jika benar, program menampilkan C.

5. **`elif nilai_2011 >= 50:`**
   Memeriksa apakah nilai 50 atau lebih. Jika benar, program menampilkan D.

6. **`else:`**
   Digunakan jika semua kondisi sebelumnya tidak terpenuhi. Artinya, nilai yang dimasukkan kurang dari 50, sehingga program menampilkan E.

```python
# 1. Header dan input data pengunjung
print("=== SISTEM LOKET ALPRO ADVENTURE PARK ===")

nama_pengunjung_2011 = input("Masukkan Nama Pengunjung : ")
umur_2011 = int(input("Input Umur Anda : "))
simC_2011 = input("Apakah Anda Sudah Punya SIM C (y/t) : ").strip().lower()[0]

print("\nPilihan Paket Wahana (1-5)")
print("1. Safari Rimba          (Rp 50,000)")
print("2. Arung Jeram           (Rp 75,000)")
print("3. Motor ATV Ekstrim     (Rp 120,000)")
print("4. Roller Coaster Kilat  (Rp 100,000)")
print("5. All-Access VIP        (Rp 220,000)")

# 2. Pemilihan Paket Wahana (match-case)
while True:
    input_tiket_2011 = int(input("Masukkan nomor paket (1-5)     : "))

    match input_tiket_2011:
        case 1:
            jenis_wahana_2011 = "Wahana Safari Rimba"
            harga_2011 = 50000
            break
        case 2:
            jenis_wahana_2011 = "Wahana Arung Jeram"
            harga_2011 = 75000
            break
        case 3:
            jenis_wahana_2011 = "Wahana Motor ATV Ekstrim"
            harga_2011 = 120000
            break
        case 4:
            jenis_wahana_2011 = "Wahana Roller Coaster Kilat"
            harga_2011 = 100000
            break
        case 5:
            jenis_wahana_2011 = "Wahana All-Access VIP"
            harga_2011 = 220000
            break
        case _:
            print("\nNomor Paket Tidak Valid. Silakan Pilih Kembali!")

# 3. Jumlah tiket dan status tambahan
while True:
    jumlah_tiket_2011 = int(input("Masukkan Jumlah Tiket     : "))
    if (jumlah_tiket_2011 > 0):
        break
    else:
        print("\nJumlah Tiket Tidak Valid. Silahkan Masukkan Kembali!")

is_member_2011 = input("Apakah Anda Member? (y/t) : ").strip().lower()[0]
is_kode_promo_2011 = input("Apakah Kode Promo Valid? (y/t) : ").strip().lower()[0]

is_kondisi_2011 = True
total_diskon_2011 = 0

# 4. Validasi kelayakan pengendara wahana
print("\n--- KELAYAKAN PENGENDARA WAHANA ---")
if (input_tiket_2011 == 3):
    if umur_2011 >= 17 and simC_2011 == "y":
        print(f"STATUS AKSES: Anda Sudah Dewasa dan Boleh Mengendarai {jenis_wahana_2011} Sendiri.")
    elif umur_2011 >= 17 and simC_2011 != "y":
        print(f"STATUS AKSES: Anda sudah dewasa tetapi tidak boleh bawa {jenis_wahana_2011} (wajib didampingi instruktur).")
    elif umur_2011 < 17 and simC_2011 =="y":
        print("Identitas tidak valid: Belum cukup umur memiliki SIM.")
    else:
        print("Anda belum cukup umur dan tidak boleh bawa {jenis_wahana_2011}.")
        is_kondisi_2011 = False

else:
    if (umur_2011 >= 10):
         print(f"STATUS AKSES : Batas Umur Tercukupi dan Anda Boleh Menaiki {jenis_wahana_2011}.")
    else:
        print(f"STATUS AKSES : Anda Tidak Cukup Umur dan Tidak Boleh Menaiki {jenis_wahana_2011}.")
        is_kondisi_2011 = False



# 5. Akumulasi diskon dan rincian pembayaran
if is_kondisi_2011:
    if (harga_2011 * jumlah_tiket_2011 >= 200000):
        total_diskon_2011 += 10
    if (is_member_2011 == "y"):
        total_diskon_2011 += 5
    if (is_kode_promo_2011 == "y"):
        total_diskon_2011 += 15
    if (jumlah_tiket_2011 >= 5):
        total_diskon_2011 += 5

    print("\n--- Rincian Pembayaran ---")
    print(f"Subtotal Belanja : Rp {harga_2011 * jumlah_tiket_2011:,.0f}")
    print(f"Total Diskon     : {total_diskon_2011}% (Rp {(harga_2011 * jumlah_tiket_2011) * (total_diskon_2011 / 100):,.0f})")
    print(f"Total Bayar       : Rp {(harga_2011 * jumlah_tiket_2011) - ((harga_2011 * jumlah_tiket_2011) * (total_diskon_2011 / 100)):,.0f}")
    if ((harga_2011 * jumlah_tiket_2011) - ((harga_2011 * jumlah_tiket_2011) * (total_diskon_2011 / 100)) > 300000):
            print("Selamat! Anda berhak mendapatkan Souvenir Gratis.")
    print("Catatan Layanan  : Terima kasih telah berkunjung.")


# 6. Penutup program
print("\nProgram Selesai")
```

**Penjelasan:**

1. **Header dan Input Data Pengunjung**

   * `print()` digunakan untuk menampilkan judul **Sistem Loket Alpro Adventure Park**.
   * `input()` digunakan untuk memasukkan nama pengunjung dan umur.
   * `int()` mengubah input umur menjadi tipe data integer.
   * `strip().lower()[0]` digunakan untuk membersihkan input SIM C dan mengambil karakter pertama, yaitu `y` atau `t`.

2. **Menampilkan Pilihan Paket Wahana**

   * Program menampilkan 5 pilihan wahana beserta harga tiketnya.
   * Setiap wahana memiliki nomor dan harga yang berbeda.
   * Pengunjung dapat memilih paket berdasarkan nomor yang tersedia.

3. **Pemilihan Paket dengan `match-case`**

   * `match-case` digunakan untuk menentukan wahana berdasarkan nomor paket yang dipilih.
   * `case 1` sampai `case 5` menentukan jenis wahana dan harga tiket.
   * `break` digunakan untuk menghentikan perulangan ketika pilihan sudah valid.
   * `case _` digunakan jika nomor yang dimasukkan tidak tersedia.

4. **Validasi Jumlah Tiket**

   * Program meminta pengguna memasukkan jumlah tiket.
   * Perulangan `while` digunakan agar pengguna dapat memasukkan kembali jumlah tiket jika tidak valid.
   * Kondisi `jumlah_tiket_2011 > 0` memastikan jumlah tiket harus lebih dari nol.

5. **Input Status Member dan Kode Promo**

   * Program meminta informasi apakah pengunjung merupakan member.
   * Program juga meminta informasi mengenai validitas kode promo.
   * Input diubah menjadi huruf kecil menggunakan `lower()` agar `Y` dan `y` dapat diproses dengan cara yang sama.

6. **Validasi Kelayakan Pengendara Wahana**

   * Program menggunakan percabangan `if-elif-else` untuk menentukan apakah pengunjung memenuhi syarat menaiki wahana.
   * Untuk Motor ATV Ekstrim, pengunjung harus berusia minimal 17 tahun dan memiliki SIM C.
   * Jika belum memenuhi syarat, nilai `is_kondisi_2011` diubah menjadi `False`.
   * Untuk wahana lainnya, pengunjung harus berusia minimal 10 tahun.

7. **Perhitungan Diskon**

   * Program menghitung diskon berdasarkan beberapa kondisi.
   * Diskon **10%** diberikan jika total belanja minimal Rp200.000.
   * Diskon **5%** diberikan jika pengunjung merupakan member.
   * Diskon **15%** diberikan jika memiliki kode promo yang valid.
   * Diskon tambahan **5%** diberikan jika membeli minimal 5 tiket.
   * Operator `+=` digunakan untuk menjumlahkan persentase diskon.

8. **Rincian Pembayaran**

   * Program menghitung subtotal berdasarkan harga tiket dikalikan jumlah tiket.
   * Program menghitung total diskon dalam bentuk persentase dan nominal rupiah.
   * Total pembayaran dihitung dengan mengurangi subtotal dengan jumlah diskon.
   * Format `:,.0f` digunakan agar angka rupiah lebih mudah dibaca.

9. **Penutup Program**

      * Setelah seluruh proses selesai, program menampilkan pesan **"Program Selesai"**.
       * Pesan tersebut menandakan bahwa seluruh proses transaksi telah selesai dilakukan.
